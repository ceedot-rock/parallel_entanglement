Import MWExperimentData to quickly process any .mat data file

```python
from MWExperimentData import MWExperimentData
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
filename = "example_data_125iMQC_L1_C0_111824.mat" #provide the full path to the data file location
exp = MWExperimentData(filename)
```

Default data analysis averages photon counts over shots and scales to a probablilty. The scaling is based on using the first two lines of any calibrated experiment to measure the $|0\rangle$ and $|1\rangle$ electron PL levels to then adjust the rest of the lines.

```python
fig,ax = plt.subplots(1,1,figsize=(5,3),constrained_layout=True)
exp.plot_data(ax,color="tab:orange",capsize=2,markersize=4,label=r"$q_1$ MQC data") #ax.errorbar kwargs can be passed directly to this, with a few examples shown
plt.show()
```

If further data manipulation or custom plotting is needed, the `MWExperimentData` instance creates helper variables `self.x_data`, `self.y_data`, `self.y_data_err`. These are the variables plotted above. Note the `y` data and errors are two dimensional arrays. The first index controls which DAQ counter channel is used. The variable `self.APD_DAQ_ctr` indexs the main data channel. Other channels are used for testing and line identification purposes. No data from other DAQ counters is plotted in the paper.
